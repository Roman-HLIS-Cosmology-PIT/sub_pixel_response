```python
import numpy as np
from sub_pixel_response.process_image import *
from astropy.io import fits
import matplotlib.pyplot as plt
```

```python
# open the image from Kara's simulation
myim = fits.open("/users/PAS2340/karadiludovico/mytest_fits/myTestPolyOct113.fits")
im = myim[0].data
```

```python
# Set some offsets up, let's try an offset of 1 in the x direction.
offsets = generate_offset_array([1, 1, 0, 0, 1 / 12 * (1 - 1 / 36), 1 / 12 * (1 - 1 / 36)])
```

```python
# Now we can apply the offsets to our image using the processImage function
offset_image = process_image(im, offsets)
fits.PrimaryHDU(offset_image).writeto(
    "/users/PAS2340/karadiludovico/sub_pixel_response/examples/offset_image_JA_7_27.fits", overwrite=True
)
```

```python
from mpl_toolkits.axes_grid1 import make_axes_locatable

fig, axes = plt.subplots(1, 3, figsize=(30, 10))

im0 = axes[0].imshow((im[:3072, :3072]), vmin=0, vmax=1e3)
axes[0].set_title("Original Image")
divider0 = make_axes_locatable(axes[0])
cax0 = divider0.append_axes("bottom", size="5%", pad=0.5)
fig.colorbar(im0, cax=cax0, orientation="horizontal")

im1 = axes[1].imshow(
    np.mean(im[:3072, :3072].reshape((512, 6, 512, 6)), axis=(1, 3)) - offset_image[:512, :512] / 36,
    vmin=-10,
    vmax=10,
    cmap="viridis",
)
axes[1].set_title("Original Image - Offset Image")
divider1 = make_axes_locatable(axes[1])
cax1 = divider1.append_axes("bottom", size="5%", pad=0.5)
fig.colorbar(im1, cax=cax1, orientation="horizontal")

im2 = axes[2].imshow(offset_image[:512, :512] / 36, vmin=0, vmax=1e3)
axes[2].set_title("Offset Image")
divider2 = make_axes_locatable(axes[2])
cax2 = divider2.append_axes("bottom", size="5%", pad=0.5)
fig.colorbar(im2, cax=cax2, orientation="horizontal")
```
